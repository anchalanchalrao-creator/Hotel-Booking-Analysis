```python
# Section A: Data Loading, Inspection and Preprocessing
# Question 1: Display first 10 and last 10 records

"""
Project: Hotel Booking Demand, Customer Behaviour and Cancellation Analysis
Author: Anchal
Description: End-to-end Exploratory Data Analysis (EDA) notebook.
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Global plot configuration
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['font.sans-serif'] = ['DejaVu Sans']
plt.rcParams['figure.dpi'] = 100

# Load the dataset
df = pd.read_csv("hotel_bookings.csv")

# Display first 10 records
display(df.head(10))

# Display last 10 records
display(df.tail(10))
```

```python

# Display all column names
print("Column Names:")
print(df.columns.tolist())

# Display data types
print("\nData Types:")
print(df.dtypes)

# Display non-null counts
print("\nNon-Null Counts:")
print(df.count())
```

```python

# Question 4: Check whether both hotel types exist

print("Hotel types available:")
print(df['hotel'].unique())

print("\nNumber of records by hotel type:")
print(df['hotel'].value_counts())
```

```python

# Question 5: Count observations for each hotel type

hotel_counts = df['hotel'].value_counts()

print("Number of bookings by hotel type:")
display(hotel_counts)


# Calculate percentage of total bookings

hotel_percentage = df['hotel'].value_counts(normalize=True) * 100

hotel_summary = pd.DataFrame({
    'Booking Count': hotel_counts,
    'Percentage (%)': hotel_percentage
})

display(hotel_summary.round(2))
```

```python

# Question 6: Identify different variable categories

# Binary variables
binary_columns = [
    'is_canceled',
    'is_repeated_guest'
]

# Date-related variables
date_columns = [
    'arrival_date_year',
    'arrival_date_month',
    'arrival_date_week_number',
    'arrival_date_day_of_month',
    'reservation_status_date'
]

# Categorical variables
categorical_columns = [
    'hotel', 'meal', 'country', 'market_segment',
    'distribution_channel', 'reserved_room_type',
    'assigned_room_type', 'deposit_type',
    'customer_type', 'reservation_status',
    'agent', 'company'
]

# Numerical variables (excluding binary and date-related columns)
numerical_columns = [
    col for col in df.select_dtypes(include=np.number).columns
    if col not in binary_columns + date_columns
]

print("Numerical Variables:", numerical_columns)
print("\nCategorical Variables:", categorical_columns)
print("\nBinary Variables:", binary_columns)
print("\nDate-related Variables:", date_columns)
```

```python

# Create a data dictionary

data_dictionary = {
    'hotel': 'Categorical',
    'is_canceled': 'Binary',
    'lead_time': 'Numerical',
    'arrival_date_year': 'Date-related',
    'arrival_date_month': 'Date-related',
    'arrival_date_week_number': 'Date-related',
    'arrival_date_day_of_month': 'Date-related',
    'stays_in_weekend_nights': 'Numerical',
    'stays_in_week_nights': 'Numerical',
    'adults': 'Numerical',
    'children': 'Numerical',
    'babies': 'Numerical',
    'meal': 'Categorical',
    'country': 'Categorical',
    'market_segment': 'Categorical',
    'distribution_channel': 'Categorical',
    'is_repeated_guest': 'Binary',
    'previous_cancellations': 'Numerical',
    'previous_bookings_not_canceled': 'Numerical',
    'reserved_room_type': 'Categorical',
    'assigned_room_type': 'Categorical',
    'booking_changes': 'Numerical',
    'deposit_type': 'Categorical',
    'agent': 'Categorical',
    'company': 'Categorical',
    'days_in_waiting_list': 'Numerical',
    'customer_type': 'Categorical',
    'adr': 'Numerical',
    'required_car_parking_spaces': 'Numerical',
    'total_of_special_requests': 'Numerical',
    'reservation_status': 'Categorical',
    'reservation_status_date': 'Date-related'
}

# Convert dictionary into a DataFrame
data_dictionary_df = pd.DataFrame(
    list(data_dictionary.items()),
    columns=['Column Name', 'Variable Type']
)

display(data_dictionary_df)
```

```python

# Question 7: Calculate missing values in each column

missing_counts = df.isnull().sum()

print("Missing Values in Each Column:")
display(missing_counts)
```

```python

# Calculate missing-value percentages

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Count': missing_counts,
    'Missing Percentage (%)': missing_percentage
})

# Display columns containing missing values
missing_summary = missing_summary[
    missing_summary['Missing Count'] > 0
].sort_values(by='Missing Percentage (%)', ascending=False)

display(missing_summary.round(2))


# Visualize missing-value distribution

plt.figure(figsize=(10, 5))

sns.barplot(
    x=missing_summary.index,
    y=missing_summary['Missing Percentage (%)']
)

plt.title('Missing Values by Column')
plt.xlabel('Columns')
plt.ylabel('Missing Values (%)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

```

```python

# Question 8: Inspect missing values in selected columns

columns_to_check = ['children', 'country', 'agent', 'company']

for column in columns_to_check:
    print(f"\nMissing values in {column}:")
    display(df[df[column].isnull()][column].head(10))
```

```python

# Examine related booking information for missing values

for column in columns_to_check:
    print(f"\nMissing {column} - Booking details:")

    display(
        df.loc[
            df[column].isnull(),
            ['hotel', 'market_segment', 'distribution_channel',
             'customer_type', 'is_canceled']
        ].head(10)
    )
```

```python

# Create a missing-value inspection report

inspection_report = pd.DataFrame({
    'Missing Count': df[columns_to_check].isnull().sum(),
    'Missing Percentage (%)':
        df[columns_to_check].isnull().mean() * 100
})

inspection_report['Possible Explanation'] = [
    'Guest count information was not recorded.',
    'Guest country information is unavailable.',
    'Booking agent information was not recorded or applicable.',
    'Company information may not apply to individual bookings.'
]

display(inspection_report.round(3))
```

```python

# Question 9: Identify duplicate records

duplicate_count = df.duplicated().sum()

print("Total duplicate records:", duplicate_count)
print("Total unique records:", df.drop_duplicates().shape[0])


# Display duplicate records

duplicates = df[df.duplicated(keep=False)]

print("Total records involved in duplication:", len(duplicates))

display(duplicates.head(10))
```

```python

# Examine characteristics of duplicate records

if not duplicates.empty:

    duplicate_summary = duplicates[
        ['hotel', 'is_canceled', 'lead_time',
         'arrival_date_year', 'arrival_date_month',
         'adults', 'children', 'adr',
         'market_segment', 'customer_type']
    ]

    display(duplicate_summary.describe(include='all').T)

    print("\nDuplicate records by hotel type:")
    display(duplicates['hotel'].value_counts())


# Verify exact duplication

print(
    "Are all identified duplicate records identical to an existing record?",
    duplicates.duplicated().sum() == len(duplicates) - df.duplicated().sum()
)

```

```python

# Question 10: Check negative stay duration

stay_columns = [
    'stays_in_weekend_nights',
    'stays_in_week_nights'
]

for column in stay_columns:
    print(f"Negative values in {column}:",
          (df[column] < 0).sum())


# Check negative guest counts

guest_columns = ['adults', 'children', 'babies']

for column in guest_columns:
    print(f"Negative values in {column}:",
          (df[column] < 0).sum())


# Check bookings with zero total guests

zero_guest_bookings = df[
    (df['adults'] + df['children'] + df['babies']) == 0
]

print("Bookings with zero guests:",
      len(zero_guest_bookings))

display(zero_guest_bookings.head())


# Inspect lead time distribution

print("Maximum lead time:", df['lead_time'].max())

print("Lead time above 365 days:",
      (df['lead_time'] > 365).sum())

print("Lead time above 1000 days:",
      (df['lead_time'] > 1000).sum())



# Check negative and zero ADR values

print("Negative ADR values:", (df['adr'] < 0).sum())

print("Zero ADR values:", (df['adr'] == 0).sum())

print("Minimum ADR:", df['adr'].min())
print("Maximum ADR:", df['adr'].max())



# Create a validation summary

validation_summary = pd.DataFrame({
    'Validation Check': [
        'Negative stay duration',
        'Negative guest counts',
        'Zero guest bookings',
        'Lead time above 365 days',
        'Negative ADR',
        'Zero ADR'
    ],
    'Number of Records': [
        sum((df[col] < 0).sum() for col in stay_columns),
        sum((df[col] < 0).sum() for col in guest_columns),
        len(zero_guest_bookings),
        (df['lead_time'] > 365).sum(),
        (df['adr'] < 0).sum(),
        (df['adr'] == 0).sum()
    ]
})

display(validation_summary)

```

```python

# Question 11: Convert arrival date components into datetime

# Convert month names into numerical month values
df['arrival_month_num'] = pd.to_datetime(
    df['arrival_date_month'],
    format='%B'
).dt.month

# Display the converted month values
display(
    df[['arrival_date_month', 'arrival_month_num']].head(10)
)


# Combine year, month and day into a single datetime column

df['arrival_date'] = pd.to_datetime({
    'year': df['arrival_date_year'],
    'month': df['arrival_month_num'],
    'day': df['arrival_date_day_of_month']
})

# Display the original components and new date column
display(
    df[[
        'arrival_date_year',
        'arrival_date_month',
        'arrival_date_day_of_month',
        'arrival_date'
    ]].head(10)
)


# Verify datetime conversion

print("Arrival date data type:", df['arrival_date'].dtype)

print("\nEarliest arrival date:", df['arrival_date'].min())
print("Latest arrival date:", df['arrival_date'].max())

print("\nMissing arrival dates:", df['arrival_date'].isna().sum())



# Verify datetime conversion

print("Arrival date data type:", df['arrival_date'].dtype)

print("\nEarliest arrival date:", df['arrival_date'].min())
print("Latest arrival date:", df['arrival_date'].max())

print("\nMissing arrival dates:", df['arrival_date'].isna().sum())

```

```python

# Question 12: Check original data type

print("Original data type:",
      df['reservation_status_date'].dtype)

display(df[['reservation_status_date']].head(10))


# Convert reservation status date into datetime format

df['reservation_status_date'] = pd.to_datetime(
    df['reservation_status_date'],
    errors='coerce'
)

print("Conversion completed successfully!")


# Verify the converted column

print("New data type:",
      df['reservation_status_date'].dtype)

print("\nEarliest reservation status date:",
      df['reservation_status_date'].min())

print("\nLatest reservation status date:",
      df['reservation_status_date'].max())

print("\nMissing values:",
      df['reservation_status_date'].isnull().sum())


```

```python

# Question 13: Calculate total nights

df['total_nights'] = (
    df['stays_in_weekend_nights'] +
    df['stays_in_week_nights']
)

# Display relevant columns
display(
    df[[
        'stays_in_weekend_nights',
        'stays_in_week_nights',
        'total_nights'
    ]].head(10)
)


# Statistical summary of total nights

print("Total Nights Summary:")
display(df['total_nights'].describe())


# Check bookings with zero or unusually high stay duration

print("Bookings with zero nights:",
      (df['total_nights'] == 0).sum())

print("Bookings with more than 30 nights:",
      (df['total_nights'] > 30).sum())

```

```python

# Question 14: Calculate total guests

df['total_guests'] = (
    df['adults'] +
    df['children'] +
    df['babies']
)

# Display relevant columns
display(
    df[[
        'adults',
        'children',
        'babies',
        'total_guests'
    ]].head(10)
)


# Statistical summary of total guests

print("Total Guests Summary:")
display(df['total_guests'].describe())


# Check bookings with zero guests

zero_guest_count = (df['total_guests'] == 0).sum()

print("Bookings with zero guests:", zero_guest_count)

# Check bookings with unusually large guest counts

print(
    "Bookings with more than 10 guests:",
    (df['total_guests'] > 10).sum()
)
```

```python
#Section B – Univariate Analysis: Booking and Customer Characteristics.
# Question 15: Descriptive statistics of lead_time

lead_time_stats = {
    'Mean': df['lead_time'].mean(),
    'Median': df['lead_time'].median(),
    'Mode': df['lead_time'].mode()[0],
    'Standard Deviation': df['lead_time'].std(),
    'Variance': df['lead_time'].var(),
    'Minimum': df['lead_time'].min(),
    'Maximum': df['lead_time'].max()
}

lead_time_summary = pd.DataFrame(
    lead_time_stats.items(),
    columns=['Statistical Measure', 'Value']
)

display(lead_time_summary.round(2))


# Additional descriptive summary

df['lead_time'].describe()

```

```python

# Question 16: Distribution of lead_time

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x='lead_time',
    bins=50,
    kde=True,
    color='steelblue'
)

plt.title('Distribution of Hotel Booking Lead Time')
plt.xlabel('Lead Time (Days)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()


# Calculate skewness of lead_time

lead_time_skewness = df['lead_time'].skew()

print("Skewness of Lead Time:", round(lead_time_skewness, 3))
if lead_time_skewness > 0:
    print("The distribution is positively (right) skewed.")
elif lead_time_skewness < 0:
    print("The distribution is negatively (left) skewed.")
else:
    print("The distribution is approximately symmetrical.")

```

```python

# Question 17: Descriptive statistics of ADR

adr_stats = {
    'Mean': df['adr'].mean(),
    'Median': df['adr'].median(),
    'Mode': df['adr'].mode()[0],
    'Standard Deviation': df['adr'].std(),
    'Variance': df['adr'].var(),
    'Minimum': df['adr'].min(),
    'Maximum': df['adr'].max()
}

adr_summary = pd.DataFrame(
    adr_stats.items(),
    columns=['Statistical Measure', 'Value']
)

display(adr_summary.round(2))


# Histogram and KDE of ADR

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x='adr',
    bins=50,
    kde=True
)

plt.title('Distribution of Average Daily Rate (ADR)')
plt.xlabel('Average Daily Rate')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()


# Boxplot to identify ADR outliers

plt.figure(figsize=(10, 3))

sns.boxplot(x=df['adr'])

plt.title('Boxplot of Average Daily Rate (ADR)')
plt.xlabel('ADR')

plt.tight_layout()
plt.show()


# Calculate ADR skewness

adr_skewness = df['adr'].skew()

print("ADR Skewness:", round(adr_skewness, 3))

if adr_skewness > 0:
    print("ADR distribution is positively skewed.")
elif adr_skewness < 0:
    print("ADR distribution is negatively skewed.")
else:
    print("ADR distribution is approximately symmetrical.")



```

```python

# Question 18: Calculate ADR percentiles

percentile_values = [5, 25, 50, 75, 90, 95, 99]

adr_percentiles = df['adr'].quantile(
    [p / 100 for p in percentile_values]
)

adr_percentile_summary = pd.DataFrame({
    'Percentile': percentile_values,
    'ADR Value': adr_percentiles.values
})

display(adr_percentile_summary.round(2))


# Plot ADR percentile distribution

plt.figure(figsize=(10, 5))

sns.barplot(
    data=adr_percentile_summary,
    x='Percentile',
    y='ADR Value'
)

plt.title('ADR Distribution Across Percentiles')
plt.xlabel('Percentile (%)')
plt.ylabel('Average Daily Rate (ADR)')

plt.tight_layout()
plt.show()

```

```python

# Question 19: Descriptive statistics of stay duration

stay_summary = df[
    ['stays_in_week_nights', 'stays_in_weekend_nights']
].describe().T

display(stay_summary.round(2))


# Calculate average weekday and weekend stays

weekday_avg = df['stays_in_week_nights'].mean()
weekend_avg = df['stays_in_weekend_nights'].mean()

print("Average weekday nights:", round(weekday_avg, 2))
print("Average weekend nights:", round(weekend_avg, 2))

if weekday_avg > weekend_avg:
    print("Weekday stays are more common on average.")
elif weekend_avg > weekday_avg:
    print("Weekend stays are more common on average.")
else:
    print("Both stay durations are equal on average.")


# Compare weekday and weekend stay distributions

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.histplot(
    data=df,
    x='stays_in_week_nights',
    bins=20,
    ax=axes[0]
)
axes[0].set_title('Distribution of Weekday Stays')
axes[0].set_xlabel('Week Nights')

sns.histplot(
    data=df,
    x='stays_in_weekend_nights',
    bins=15,
    ax=axes[1]
)
axes[1].set_title('Distribution of Weekend Stays')
axes[1].set_xlabel('Weekend Nights')

plt.tight_layout()
plt.show()


# Bar chart of average stay duration

stay_averages = pd.Series({
    'Weekday Nights': weekday_avg,
    'Weekend Nights': weekend_avg
})

plt.figure(figsize=(7, 5))

sns.barplot(
    x=stay_averages.index,
    y=stay_averages.values
)

plt.title('Average Weekday vs Weekend Stay Duration')
plt.xlabel('Stay Type')
plt.ylabel('Average Number of Nights')

plt.tight_layout()
plt.show()



```

```python

# Question 20: Descriptive statistics of guest composition

guest_columns = ['adults', 'children', 'babies']

guest_summary = df[guest_columns].describe().T

display(guest_summary.round(2))


# Calculate mode of each guest category

for column in guest_columns:
    print(f"\nMost common value for {column}:")
    print(df[column].mode().tolist())


# Plot distribution of adults, children and babies

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for column, ax in zip(guest_columns, axes):
    sns.countplot(data=df, x=column, ax=ax)
    ax.set_title(f'Distribution of {column.capitalize()}')
    ax.set_xlabel(column.capitalize())
    ax.set_ylabel('Number of Bookings')

plt.tight_layout()
plt.show()


# Calculate average guest counts

guest_averages = df[guest_columns].mean()

print("Average guest composition:")
display(guest_averages.round(2).to_frame('Average Count'))
```

```python
# Define conditions for identifying unusual ADR values
zero_adr = df['adr'] == 0
negative_adr = df['adr'] < 0
high_adr = df['adr'] > 500

# Count bookings in each category
adr_quality = pd.DataFrame({
    'Category': [
        'Zero ADR',
        'Negative ADR',
        'Unusually High ADR (>500)'
    ],
    'Booking_Count': [
        zero_adr.sum(),
        negative_adr.sum(),
        high_adr.sum()
    ]
})

# Calculate percentage of total bookings
adr_quality['Percentage'] = (
    adr_quality['Booking_Count'] / len(df) * 100
)

# Display the results
display(adr_quality.round(2))

# Calculate the number and percentage of bookings with any flagged ADR
flagged_adr = zero_adr | negative_adr | high_adr

flagged_count = flagged_adr.sum()
flagged_percentage = flagged_count / len(df) * 100

print(f"Total bookings: {len(df)}")
print(f"Bookings with flagged ADR values: {flagged_count}")
print(f"Percentage of flagged bookings: {flagged_percentage:.2f}%")

# Display descriptive statistics for flagged ADR records
display(
    df.loc[flagged_adr, ['hotel', 'adr', 'is_canceled',
                         'total_nights']].head(10)
)

# Visualize the number of bookings in each ADR category
plt.figure(figsize=(9, 5))

sns.barplot(
    data=adr_quality,
    x='Category',
    y='Booking_Count',
    errorbar=None
)

plt.title('Bookings with Zero, Negative or Unusually High ADR')
plt.xlabel('ADR Category')
plt.ylabel('Number of Bookings')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

# Examine the overall ADR distribution using a boxplot
plt.figure(figsize=(8, 5))

sns.boxplot(x=df['adr'])

plt.title('Overall ADR Distribution and Outliers')
plt.xlabel('Average Daily Rate (ADR)')
plt.tight_layout()
plt.show()
```
